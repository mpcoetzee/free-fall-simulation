```python
import matplotlib.pyplot as plt
import numpy as np

# Set up the parameters for the simulation
displacement = 0.0
initialVelocity = 0.0
acceleration = 9.81
time = 10.0
timeStep = 0.1
currentTime = 0.0
currentVelocity = initialVelocity
finalDisplacement = 0.0
finalVelocity = 0.0
```

```python
# Initialize lists to record time and displacement
time_values = []
displacement_values = []
velocity_values = []

# Record values at each time step
while currentTime <= time:
    time_values.append(currentTime)
    
    displacement = (1/2 * acceleration) * (currentTime ** 2)
    displacement_values.append(displacement)
    
    currentVelocity = acceleration * currentTime
    velocity_values.append(currentVelocity)
    
    currentTime += timeStep

# Convert lists to numpy arrays for consistency
time_values = np.array(time_values)
displacement_values = np.array(displacement_values)
velocity_values = np.array(velocity_values)
```

```python
#Final Value Calculation
finalDisplacement = (initialVelocity * time) + (1/2) * acceleration * (time ** 2)
print(displacement)

finalVelocity = initialVelocity + (acceleration * time)
print(finalVelocity)
```

```python
plt.plot(time_values, displacement_values)

plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Displacement vs Time')

plt.show()
```

```python
plt.plot(time_values, velocity_values)

plt.xlabel('Time (s)')
plt.ylabel('velocity (m/s)')
plt.title('Velocity vs Time')

plt.show()
```
